```python
harmonized = pd.read_csv("harmonized_comparison.csv")

summary = pd.read_csv("harmonized_summary.csv")

performance = pd.read_csv("model_performance.csv")
```

```python
harmonized.head()
```

```python
# Average Screen Time Across Age Groups

plt.figure(figsize=(6,5))

sns.barplot(

    data=summary,

    x="age_group",

    y="Average_Screen_Time",

    hue="source"

)

plt.title("Average Daily Screen Time by Age Group")

plt.xlabel("Age Group")

plt.ylabel("Hours")

plt.xticks(rotation=15)

plt.tight_layout()

plt.savefig(

    "results/screen_time_by_age_group.png",

    dpi=300

)

plt.show()
```

```python
# Percentage of Sleep-Affected Individuals

summary["Percentage_Sleep_Affected"] *= 100

plt.figure(figsize=(6,5))

sns.barplot(

    data=summary,

    x="age_group",

    y="Percentage_Sleep_Affected",

    hue="source"

)

plt.title("Sleep Problems Across Age Groups")

plt.ylabel("Percentage (%)")

plt.tight_layout()

plt.savefig(

    "results/sleep_affected_percentage.png",

    dpi=300

)

plt.show()
```

```python
# Screen Time Distribution

plt.figure(figsize=(6,5))

sns.boxplot(
    data=harmonized,
    x="age_group",
    y="screen_time_hours",
    hue="source"
)

plt.title("Distribution of Daily Screen Time")
plt.tight_layout()
plt.savefig(
    "results/screen_time_distribution.png",
    dpi=300
)

plt.show()
```

```python
# Sleep-Affected Distribution

plt.figure(figsize=(6,5))
sns.countplot(
    data=harmonized,
    x="age_group",
    hue="sleep_affected"
)

plt.title("Sleep Problems by Age Group")
plt.tight_layout()
plt.savefig(
    "results/sleep_problem_distribution.png",
    dpi=300
)

plt.show()
```

```python
# Adult vs Kids Comparison

plt.figure(figsize=(6,5))
sns.violinplot(
    data=harmonized,
    x="source",
    y="screen_time_hours"
)

plt.title("Adult vs Kids Screen Time")
plt.tight_layout()
plt.savefig(
    "results/adult_vs_kids_violin.png",
    dpi=300
)

plt.show()
```

## Conclusion

This study investigated the relationship between screen exposure and sleep health using complementary datasets representing children, teenagers, and adults.

After harmonizing the datasets, exploratory analysis, machine learning, and cross-age comparisons demonstrated that increased screen exposure is associated with a greater likelihood of sleep-related issues. While the datasets differ in their populations and sleep outcome definitions, the harmonized analysis provides a broader perspective on how digital device usage influences sleep across different stages of life.

The project demonstrates the application of data preprocessing, exploratory analysis, machine learning, and research-oriented data interpretation to address a contemporary public health problem.
