# Feature Engineering

To compare the two datasets meaningfully, common features are created.

The following harmonized variables are generated:

- Age Group
- Screen Time (hours)
- Sleep Affected (Binary)

These common variables enable cross-age-group analysis while preserving the integrity of the original datasets.

```python
# Create Common Age Groups

AGE_BINS = [7, 12, 17, 29, 45]

AGE_LABELS = [
    "8-12 (Pre-teen)",
    "13-17 (Teen)",
    "18-29 (Young Adult)",
    "30-45 (Adult)"
]

adult_df["age_group"] = pd.cut(
    adult_df["Age"],
    bins=AGE_BINS,
    labels=AGE_LABELS
)

kids_df["age_group"] = pd.cut(
    kids_df["Age"],
    bins=AGE_BINS,
    labels=AGE_LABELS
)

print("Age Groups Created Successfully")
```

```python
adult_df["age_group"].value_counts().sort_index()
```

```python
kids_df["age_group"].value_counts().sort_index()
```

```python
adult_df["screen_time_hours"] = adult_df["Average screen time"]

kids_df["screen_time_hours"] = kids_df["Avg_Daily_Screen_Time_hr"]

print("Screen Time Feature Created")
```

```python
adult_df["sleep_affected"] = (

    (adult_df["Sleep quality"] <= 2)

    |

    (adult_df["Sleep disorder"] == "Y")

).astype(int)
```

```python
kids_df["sleep_affected"] = (

    kids_df["Health_Impacts"]

    .fillna("")

    .str.contains(
        "Poor Sleep",
        case=False
    )

).astype(int)
```

## Harmonization

Only common variables are retained for comparison.

This enables meaningful cross-age-group analysis while avoiding invalid row-wise merging of unrelated datasets.

```python
comparison_adult = adult_df[

    [
        "age_group",
        "Gender",
        "screen_time_hours",
        "sleep_affected"
    ]

].copy()

comparison_adult["source"] = "Adult"
```

```python
comparison_kids = kids_df[

    [
        "age_group",
        "Gender",
        "screen_time_hours",
        "sleep_affected"
    ]

].copy()

comparison_kids["source"] = "Kids"
```

```python
harmonized = pd.concat(

    [
        comparison_adult,
        comparison_kids
    ],

    ignore_index=True

)

harmonized.rename(

    columns={
        "Gender":"gender"
    },

    inplace=True
)

harmonized.head()
```

```python
harmonized.groupby(

    ["source","age_group"],
    observed=True

).agg(

    Average_Screen_Time=("screen_time_hours","mean"),

    Percentage_Sleep_Affected=("sleep_affected","mean"),

    Participants=("sleep_affected","count")

)
```

```python
adult_df.to_csv(

    "results/cleaned_adult_dataset.csv",

    index=False

)

kids_df.to_csv(

    "results/cleaned_kids_dataset.csv",

    index=False

)

harmonized.to_csv(

    "results/harmonized_comparison.csv",

    index=False

)

print("Datasets Saved Successfully")
```

```python
summary = (
    harmonized.groupby(["source", "age_group"], observed=True)
    .agg(
        Average_Screen_Time=("screen_time_hours", "mean"),
        Percentage_Sleep_Affected=("sleep_affected", "mean"),
        Participants=("sleep_affected", "count"),
    )
    .reset_index()
)

summary.to_csv("results/harmonized_summary.csv", index=False)
```
